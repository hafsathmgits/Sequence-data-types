```python
my_list=[]
n=int(input("Enter the value of n:"))
for i in range (n):
  ele=int(input("Enter the element:"))
  my_list.append(ele)
print(my_list)

my_list_odd=[]
my_list_even=[]
for item in my_list:
  if item%2==0:
    my_list_even.append(item)
  else:
    my_list_odd.append(item)

print(my_list_even)
print(my_list_odd)
```
