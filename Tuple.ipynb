```python
# Relational Operators in Python Sets

# Consider the following sets
A = {1, 2}
B = {1, 2, 3, 4}
C = {1, 2}

# 1. Subset (<=)
# A is a subset of B if every element of A is present in B.
print("Subset (A <= B):", A <= B)  # True
print("Subset (A <= C):", A <= C)  # True

# 2. Proper Subset (<)
# A is a proper subset of B if every element of A is in B
# and A and B are not equal.
print("Proper subset (A < B):", A < B)  # True
print("Proper subset (A < C):", A < C)  # False

# 3. Superset (>=)
# B is a superset of A if B contains every element of A.
print("Superset (B >= A):", B >= A)  # True
print("Superset (A >= C):", A >= C)  # True

# 4. Proper Superset (>)
# B is a proper superset of A if B contains all elements of A
# and has at least one additional element.
print("Proper superset (B > A):", B > A)  # True
print("Proper superset (A > C):", A > C)  # False

# 5. Equal Sets (==)
# Two sets are equal if they contain exactly the same elements.
# The order of elements does not matter.
X = {1, 2, 3}
Y = {3, 2, 1}
print("Equal sets (X == Y):", X == Y)  # True

# 6. Not Equal Sets (!=)
# Two sets are not equal if they contain different elements.
P = {1, 2, 3}
Q = {1, 2, 4}
print("Not equal sets (P != Q):", P != Q)  # True
```

```python


print("\n========== SET METHODS ==========")

# Use fresh sets for each example so that one update operation
# does not affect the next example.
A = {1, 2, 3}
B = {3, 4, 5}

print("\nOriginal sets:")
print("A =", A)
print("B =", B)

# 1. union(): Returns a NEW set containing all elements from A and B.
result = A.union(B)
print("\n1. A.union(B) =", result)                     # {1, 2, 3, 4, 5}
print("A after union() =", A)                           # A remains unchanged

# 2. intersection(): Returns a NEW set containing common elements.
result = A.intersection(B)
print("\n2. A.intersection(B) =", result)              # {3}
print("A after intersection() =", A)                    # A remains unchanged

# 3. intersection_update(): Modifies A to keep only common elements.
A1 = {1, 2, 3}
A1.intersection_update(B)
print("\n3. After A1.intersection_update(B), A1 =", A1) # {3}

# 4. difference(): Returns a NEW set containing elements in A but not B.
A2 = {1, 2, 3}
result = A2.difference(B)
print("\n4. A2.difference(B) =", result)                # {1, 2}
print("A2 after difference() =", A2)                     # A2 remains unchanged

# 5. difference_update(): Modifies A to remove elements also present in B.
A3 = {1, 2, 3}
A3.difference_update(B)
print("\n5. After A3.difference_update(B), A3 =", A3)   # {1, 2}

# 6. symmetric_difference(): Returns a NEW set containing elements
# that are in either set, but NOT in both.
A4 = {1, 2, 3}
result = A4.symmetric_difference(B)
print("\n6. A4.symmetric_difference(B) =", result)      # {1, 2, 4, 5}
print("A4 after symmetric_difference() =", A4)           # A4 remains unchanged

# 7. symmetric_difference_update(): Modifies A to keep elements
# that are in either set, but NOT in both.
A5 = {1, 2, 3}
A5.symmetric_difference_update(B)
print("\n7. After A5.symmetric_difference_update(B), A5 =", A5) # {1, 2, 4, 5}


print("\n========== SUMMARY ==========")
print("union()                  : all elements from both sets")
print("intersection()           : common elements")
print("intersection_update()    : updates the set with common elements")
print("difference()             : elements in the first set only")
print("difference_update()      : removes common elements from the first set")
print("symmetric_difference()   : elements in either set, but not both")
print("symmetric_difference_update(): updates the set with non-common elements")

# Important:
# Methods without '_update' return a new set and do not change the original set.
# Methods with '_update' modify the original set directly.
```
