```python
# Answers

# Python Functions - Basic Programs
# First-year practice questions with solutions

# 1. Function without arguments - welcome()
def welcome():
    print("Welcome to Python Programming")


welcome()


# 2. Function without arguments - display()
def display():
    print("Name: Anu")
    print("College: ABC College")
    print("Department: Computer Science")


display()


# 3. Function with one argument - even_odd()
def even_odd(num):
    if num % 2 == 0:
        print(num, "is Even")
    else:
        print(num, "is Odd")


even_odd(10)
even_odd(7)


# 4. Function with one argument - square()
def square(num):
    print("Square =", num * num)


square(5)
square(8)


# 5. Function with two arguments - add()
def add(a, b):
    print("Sum =", a + b)


add(10, 20)
add(5, 15)


```

```python
# Python Functions - Functions with Default Arguments
# First-year practice questions with solutions

# 11. greet() - name has a default value
def greet(name="Student"):
    print("Hello", name)


greet()
greet("Anu")


# 12. calculate() - second number has a default value of 10
def calculate(a, b=10):
    print("Sum =", a + b)


calculate(5)
calculate(5, 20)


# 13. bill_amount() - quantity has a default value of 1
def bill_amount(price, quantity=1):
    total = price * quantity
    print("Total bill =", total)


bill_amount(50)
bill_amount(50, 3)


# 14. employee() - department and designation have default values
def employee(name, department="Computer Science",
             designation="Assistant Professor"):
    print("Employee Name:", name)
    print("Department:", department)
    print("Designation:", designation)
    print()


employee("Anu")
employee("Rahul", "Electronics")
employee("Meera", "Information Technology", "HOD")


# 15. power() - exponent has a default value of 2
def power(num, exponent=2):
    print("Result =", num ** exponent)


power(5)
power(2, 3)
```
